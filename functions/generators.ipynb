```python
def fruits():
    l=["apple","banana","grapes","mango"]
    for i in l:
        return i
fruits()
    
```

```python
def fruits():
    l=["apple","banana","grapes","mango"]
    for i in l:
        yield i
res=fruits()
print(list(res))
```

```python
#how to create custom iterator by using generator
l=[1,2,3,4,5,6,7,8]
def only2(l):
    for x in l:
        if x%2==0:
            yield x
g=only2(l)
```

```python
#custom define iterator
#ITER()
l=[3,4,5,6,999,8,9,4]
s1=0
it=iter(l)

for x in it:
    print(x)
    if x==999:
        break
    s1+=x
print("--------------")
s2=0

for x in it:
    print(x)
    s2+=x

print(s1-s2)
```
