```python
#Recursion : a function calling itself is called recursion
def fun(n):
    if n>0:
        print(n)
        fun(n-1)
fun(5)

```

```python
# factorial
def fun(n):
    if n==0 or n==1:
        return 1
    else:
        return n*fun(n-1)
fun(3)
```
