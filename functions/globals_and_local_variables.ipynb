```python
g=8
def fun():
    g=10
def start():
    print("before fun :",g)
    fun()
    print("after fun :",g)
start()
```

```python
#to change global value
g=8
def fun():
    global g
    g=10
def start():
    print("before fun :",g)
    fun()
    print("after fun :",g)
start()
```

```python
# GLOBAL SCOPE
x = 20

def show_global():
    print("Global x:", x)

show_global()

# LOCAL SCOPE
def show_local():   
    x = 10
    print("Local x:", x)
show_local()

# GLOBAL SCOPE
print("Global x again:", x)


```
