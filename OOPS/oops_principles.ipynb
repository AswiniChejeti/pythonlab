```python
# Class can be defined  to model any entity ,process or an idea
#object is the supermost class  wcich contain __init__,__dict__,etc
class A(object):
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def process(self):
        return self.x+self.y
obj=A(3,4)
obj.process()
```

```python
#inheritance
class Base(object):
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def add(self):
        print(self.x+self.y)

class Derived(Base):
    pass
obj=Derived(9,2)
obj.add()
```
