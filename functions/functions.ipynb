```python
#Write a function that returns the sum of numbers from 1 to n.
def sum_natural(n):
    sum1=0
    for i in range(1,n+1,1):
        sum1+=i
    return sum1
sum_natural(5)
```

```python
#Write a function that returns the factorial of a number.

def factorial(n):
    product=1
    i=1
    while i<=n:
        product*=i
        i+=1
    return product
factorial(5)

           
```

```python
#factorial
def factorial(m):
    if m==0:
        return 1
    else:
        return m*factorial(m-1)
factorial(4)

```

```python
#Write a function that checks if a number is positive, negative, or zero.
def num(n):
    if n>0:
        return "positive" 
    elif n<0:
        return "negative"
    else:
        return "zero"
res=num(-6)
print(res)
```

```python
#Write a function that counts vowels in a string.
def vowels(s):
    count=0
    vowels="aeiou"
    for ch in s:
        if ch in vowels:
            count+=1
    return count
res=vowels("hagiukrilmvioyunybxu")
print(res)
```

```python
#Write a function that reverses a string.
def rev(s):
    return s[::-1]
res=rev("apple")
print(res)
```

```python
#Write a function that returns the length of a list (don’t use len()).
def length(l):
    count=0
    for _ in l:
        count+=1
    return count
l=[23,5,67,80,566,90,97]
res=length(l)
print(res)
```

```python
#Write a function that finds the maximum element in a list.
def largest(l):
    i=0
    largest=l[0]
    while i<len(l):
        if l[i]>largest:
            largest=l[i]
        i+=1
    return largest
l=[2,4,65,5,4]
res=largest(l)
print(res)
```

```python
#Write a function that counts how many times a number appears in a list.
def repeat(l,re):
    count=0
    for i in l:
        if i==re:
            count+=1
    return count
l=[2,5,7,9,5,7,8,5]
re=5
res=repeat(l,re)
print(res)
```

```python
#Write a function to check if a number is prime.
def prime(n):
    if n<=1:
        return False
    for i in range(2,n%10):
        if n%i ==0:
            return False
    else:
        return True
print(prime(68))
```

```python
#Write a function that returns all even numbers from a list.
def even_num(l):
    ls=[]
    for i in l:
        if i%2==0:
            ls.append(i)
    return ls
l=[2,3,4,5,6]
print(even_num(l))
```

```python
#Write a function that checks if a string is a palindrome.
def palindrome(s):
    if s==s[::-1]:
        return True
    else:
        return False
print(palindrome("mAdAm"))
```

```python
#Write a function that returns the sum of digits of a number.
def sum_digits(num):
    num=str(num)
    sum1=0
    for i in num:
        sum1+=int(i)
    return sum1
print(sum_digits(123))
```

```python
def sum_digi(num):
    i=0
    sum1=0
    while num!=0:
        ld=num%10
        sum1+=ld
        num=num//10
        i+=1
    return sum1
print(sum_digi(123))
```
