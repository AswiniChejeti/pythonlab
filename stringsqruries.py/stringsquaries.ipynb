```python
#1, Reverse a string by using slicing
data="Hello World!"
print(data[::-1])
```

```python
# 2, reverse a string by using reversed function

data="Good Morning!" 
data1=reversed(data)
output=''.join(data1)
print(output)
```

```python
#3, reverse a string by using  while loop
data="Have a nice day"
i=len(data)-1
output=""
while i>=0:
    output+=data[i]
    i-=1
print(output)
```

```python
#4, write a program to reverse order of words in gien input

data="python is high level general purpose programing language"
data1=data.split()
output=' '.join(data1[::-1])
print(output)
```

```python
#5 write a program to reverse of internal content of each word

data="python programing language is easy to learn bcz easy syntax"
data1=data.split()
output=[]
for word in data1:
    output.append(word[::-1])
print(' '.join(output))

```

```python
#6, write a program to reverse internal content of every second word in string
data="python is an interpreted programming language"
data1=data.split()
i=0
output=[]
while i<len(data1):
    if i%2==0:
        output.append(data1[i][::-1])
    else:
        output.append(data1[i])
    i+=1
print(' '.join(output))
```

```python
#7, write a program to the character present at even index and odd index seperatly for the given string
data="python"
odd=""
even=""
for i,ch in enumerate(data):
    if i%2==0:
        even+=ch
    else:
        odd+=ch
print("even index characters :",even)
print("odd index charctera :",odd)

```

```python
#8,write a program to merge two string alternatively
word1="python"
word2="java"
i,j=0,0
output=""
while i<len(word1) or j<len(word2):
    if i<len(word1):
        output+=word1[i]
        i+=1
    if j<len(word2):
        output+=word2[j]
        j+=1    
print(output)
```

```python
#9,program to sort caracters of the string  first alphabet symbols follwed by digits
data="py1th2on3"
alpha=[]
digit=[]
for ch in data:
    if ch.isalpha():
        alpha.append(ch)
    elif ch.isdigit():
        digit.append(ch)
print(''.join(sorted(alpha)+sorted(digit)))

```

```python
#10, even index char as capital and odd index char as small
data="GeN AI University"
result=""
index=0
for ch in data:
    if ch==" ":
        result+=" "
    else:
        if index%2==0:
            result+=ch.upper()
        else:
            result+=ch.lower()
    index+=1
print(result)
```

```python
#11, program for the requirement
#input:a4b3c2 expected output:aaaabbbcc

data="a4b3ec2"
output=""
for i,ch in enumerate(data):
    if ch.isalpha() and data[i+1].isdigit():
        output+=ch*int(data[i+1])
    elif ch.isalpha():
        output+=ch
print(output)
```
