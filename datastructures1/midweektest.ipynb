```python
words=['eat', 'tea', 'bat', 'ate', 'tab']
output={}
for word in words:
    key=''.join(sorted(word))
    if key not in output:
        output[key]=[]
    output[key].append(word)
print(list(output.values()))
    
```

```python
data=['tea','ate','god','dog']
result={}
for data1 in data:
    anagram=''.join(sorted(data1))
    if anagram not in result:
        result[anagram]=[]
    result[anagram].append(data1)
print(result.values())
```

```python
l=['apple','boll','llob','lobl']
result={}
for l1 in l:
    sign=''.join(sorted(l1))
    if sign not in result:
        result[sign]=[]
    result[sign].append(l1)
print(list(result.items()))
```
