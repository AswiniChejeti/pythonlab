```python
#at the above code we get exception due to that our execution will be terminated due to that
l=["3","4.5","4","6.8","9.0"]
for x in l:
    try:
        y=int(x)
        print(f"{y}--------->{y*y}")
    except ValueError:
        print (f"cannot process{x}")    
```

```python
import logging

logger=logging.getLogger("ipynb")

l=["3","4.5","4","6.8","9.0"]


# when we know the type of error we directly mention the the type 
for x in l:
    try:
        y=int(x)
        print(f"{y}--------->{y*y}")
    except ValueError:
       logging.error(f"cannot process{x}")    
```

```python
#if we don't know the type of error  then we follow the bellow proceess while learning
#Exception is the mother class of all types of errors
l=["3","4.5","4","6.8","9.0"]
for x in l:
    try:
        y=int(x)
        print(f"{y}--------->{y*y}")
    except Exception as ex:
        print (f"cannot process{ex}")    
```

```python
# if we want to know the type of error for above code
l=["3","4.5","4","6.8","9.0"]
for x in l:
    try:
        y=int(x)
        print(f"{y}--------->{y*y}")
    except Exception as ex:
        print (f"cannot process{type(ex)}")    
```

```python
# while lerning print is fine ,but defintly use logging.error every where it's important while development
# else is possible when try is successfull ,otherwise any exception block is execute if try fails

l=["3","45","4","68","90"]
for x in l:
    try:
        y=int(x)
        print(f"{y}--------->{y*y}")
    except Exception as ex:
        print (f"cannot process{type(ex)}")
    else:
        print("it's cool") 



```

```python
#finally
# if try is successfull it comes to else block it's executed then finally block is executed
#if try is failure it comes to any except block is executed then also finally block is executed

l=[2,3,4,5,6,7]
n=0
for x in l:
    try:
        res=x/n
    except ZeroDivisionError as ex:
        print(f"cannot process because n is zero")
    except Exception as ex:
        print(type(ex))
    else:
        print("no error")
    finally:
        print("thank you")
```
