```python
#execution time calculation using decorator 
import time
def outer(f):
    def inner():
        start=time.time()
        r=f()
        end=time.time()
        print(end-start)
        return r
    return inner
@outer
def sum1():
    sum1=0
    for i in range(1,1000000):
        sum1+=i
    return sum1
res=sum1()
print(res)
```

```python
#in depth understanding of decorator

def count(f):
    def wrapper(n):
        print(f"fibbonacci with {n}")
        res=f(n)
        print(f"febbonacci with {n}")
        return res
    return wrapper
@count
def febbonacci(n):
    if n<=1:
        return n
    return febbonacci(n-2)+febbonacci(n-1)
res=febbonacci(4)
print(res)
```
