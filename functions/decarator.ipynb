```python
#execution time calculation using decorator 
import time
def outer(f):
    def inner():
        start=time.time()
        r=f()
        end=time.time()
        print(end-start)
        return r
    return inner
@outer
def sum1():
    sum1=0
    for i in range(1,1000000):
        sum1+=i
    return sum1
res=sum1()
print(res)
```

```python
#in depth understanding of decorator

def count(f):
    def wrapper(n):
        print(f"fibbonacci with {n}")
        res=f(n)
        print(f"febbonacci with {n}")
        return res
    return wrapper
@count
def febbonacci(n):
    if n<=1:
        return n
    return febbonacci(n-2)+febbonacci(n-1)
res=febbonacci(4)
print(res)
```

```python
def add(x,y):
    print("*"*30)
    print(x+y)
    print("*"*30)

def simple_interest(p,t,r):
    print("*"*30)
    print((p*t*r)/100)
    print("*"*30)

add(4,5)
simple_interest(100000,12,2)
```

```python
def stars(f):
    def wrapper(*args,**kwargs):
        print("*"*30)
        res=f(*args,**kwargs)
        print("*"*30)
        return res
    return wrapper


@stars
def add(x,y):
    print(x+y)

@stars
def simple_interest(p,t,r):
    print((p*t*r)/100)

add(6,7)
simple_interest(10000,13,2)
```

```python
#1,function can define inside another function
#2,function can be passed to another function
#3, function can be returned another function
```

```python
#2,function can be passed to another function
def greet():
    print("welcome python")
def fun(f):
    print("*"*20)
    f()
    print("*"*20)
fun(greet)
```

```python
#1,function can define inside another function
def fun():
    pi=3.14
    def area(r):
        a=pi*r*r
        print("area= ",a)
    print("_________________")
    area(5)
    print("___________________")
fun()
```

```python
#3, function can be returned another function
def fun():
    pi=3.14
    def area(r):
        a=pi*r*r
        print("area= ",a)
    return area

res=fun()
res(5)
```

```python
#double decarators
def stars(f):
    def wrapper(*args,**kwargs):
        print("*"*30)
        res=f(*args,**kwargs)
        print("*"*30)
        return res
    return wrapper

def dash(f):
    def wrapper(*args,**kwargs):
        print('-----------------------------')
        res=f(*args,**kwargs)
        print('------------------------------')
        return res
    return wrapper


@dash
@stars
def add(x,y):
    print(x+y)

@stars
def simple_interest(p,t,r):
    print((p*t*r)/100)

add(6,7)
simple_interest(10000,13,2)
```

```python
#time taken by function using decarator

import time

def timer(f):
    def wrapper(*args,**kwargs):
        start=time.time()
        res=f(*args,**kwargs)
        end=time.time()
        print("timetaken : ",end-start)
        return res
    return wrapper
@timer
def process(n):
    sum=0
    for i in range(1,n+1):
        sum+=i
process(10000000)
```
