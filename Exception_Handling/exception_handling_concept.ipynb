```python
#at the above code we get exception due to that our execution will be terminated due to that
l=["3","4.5","4","6.8","9.0"]
for x in l:
    try:
        y=int(x)
        print(f"{y}--------->{y*y}")
    except ValueError:
        print (f"cannot process{x}")    
```
