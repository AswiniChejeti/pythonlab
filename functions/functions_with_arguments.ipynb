```python
#nCr  ==>  n!/(r!*(n-r)!)
def fact(num):
    f=1
    for x in range (1,num+1):
        f=f*x
    return f
def ncr(n,r):
    res=fact(n)/(fact(r)*fact(n-r))
    return res
print(ncr(5,2))
```

```python
#positional arguments
def add(x,y,z):
    return x+y+z
print(add(10,20,30))


```

```python
#keyword arguments
def add(x,y,z):
    return x+y+z
print(add(5,z=23,y=90))
#add(23,z-10,34) not possible
```

```python
#Default arguments
def add(x,y,z=10):
    return x+y+z
print(add(10,20))

```

```python
#default
def add(a,b,c=5):#if value is not given then defaultly it takes as 5 for c
    return a+b+c
print(add(10,20,30))
```

```python
#variable arguments
def add(*args):
    print(type(args))
    for x in args:
        print(x)
    return sum(args)
add(1,2,3,4)
```

```python
#a tuple into parameters(parameters unpacking)
t=(1,2,3)
def add(x,y,z):
    print(x+y+z)
add(t[0],t[1],t[2])

```

```python
#Variable keyword arguments
def person(**kwargs):
    print(type(kwargs))
    print(kwargs)
person(a=10,b=20,c=90,d='apple')
    
```

```python
def emp(_id,_name,_sal,**kwargs):
    print(f"eid :{_id},ename :{_name},esal:{_sal}")
    for key,val in kwargs.items():
        print(f"{key}:{val},",end="")
    print()
print(emp(1,"raj",354000,age=23))
```

```python
d={"x":10,"y":20 , "z":30}
t=(1,2,3)
l=[12,13,12]

def fun(x,y,z):
    print(x+y+z)
fun(d['x'],d['y'],d['z'])
fun(*t)
fun(*l)
fun(**d)

```
