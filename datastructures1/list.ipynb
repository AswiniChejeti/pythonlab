```python
#4. Write a Python program to find common items from two lists.
_list4=input("enter list1 values").split(',')
list4=input("enter list2 values").split(',')
output=[]
for i in range(0,len(_list4)):#o(n2) complexity
    if _list4[i] in list4:
        output.append(_list4[i])
print(output)
```

```python
#11 Write a program to find difference between sum of even indexed and odd indexed numbers in a list of numbers. [Note:- Consider 0th index as even indexed]
list11=[2, 4, 3, 9, 13, 12, 7, 6, 1, 5]
even_sum=0
odd_sum=0
for i in range(0,len(list11)):
    if i%2==0:
        even_sum+=list11[i]
    else:
        odd_sum+=list11[i]
print(even_sum-odd_sum)
        
```

```python
#12, Write a program to print the list of numbers which are greater than the average of numbers in the following list.
list12=[2, 4, 3, 9, 13, 12, 7, 6, 1, 5]
avg=0
sum=0
output12=[]
for i in range(0,len(list12)):
    sum+=list12[i]
avg=sum/len(list12)
for j in list12:
    if j>avg:
        output12.append(j)
print(output12)

```

```python
#14,Write a program to know how many times an element occurred in the list.
list14 = [5, 10, 15, 20, 25, 50, 20]
#using dictionaries
num=int(input('enter a value :'))
dict14={}
for val in list14:
    if val in dict14:
       dict14[val]+=1
    else:
        dict14[val]=1
print(dict14.get(num,'no value present'))
#USING LIST
print(list14.count(num))
```

```python
#17, Two lists are given below. Write a program to create a third list by picking an odd-index element from the first list and even index elements from second.
# Sample Input
list17 = [3, 6, 9, 12, 15, 18, 21]
list18= [4, 8, 12, 16, 20, 24, 28]
list19=[]
for i in range(0,len(list17)):
    if i%2!=0:
        list19.append(list17[i])
for j in range(0,len(list18)):
    if j%2==0:
        list19.append(list18[j])
print(list19)
```
