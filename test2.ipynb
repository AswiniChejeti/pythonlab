```python
n1 = int(input("enter a number  : "))
n2 = n1 + 1
l2 = [n1, n2]
for roon in l2:
    factor= []
    total=0
    for i in range(1, roon+1):
        if roon % i == 0:
            factor.append(i)

    print(f"Divisors of {roon}: {factor}")

    prime_divisors = []
    for j in factor:
        if j > 1: 
            for k in range(2, int(j**0.5) + 1):
                if j % k == 0:
                    break
            else:  # runs only if loop didn't break → j is prime
                prime_divisors.append(j)
                total += j
                print(j)
        
    print(f"Running total: {total}")
```
