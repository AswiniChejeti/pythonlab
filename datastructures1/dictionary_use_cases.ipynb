```python
#counting
list1=[1, 1, 6, 2, 2, 3, 3, 4, 5, 5, 6]
dict1={}
for i in list1:
    if i in dict1:
        dict1[i]+=1
    else:
        dict1[i]=1
print(dict1)

```

```python
#Grouping
dict2={}
odd=list()
even=list()
for i in range(1,11):
    if i%2!=0:
        odd.append(i)
    else:
        even.append(i)
dict2.update({"odd":odd,"even":even})
print(dict2)
```

```python
#indexing
dict1={}
for i in range(1,11):
    dict1[i]=i
print(dict1)
```

```python
#grouping
trans=[(1237,87522),
       (1234,1000),
       (1999,34657),
       (1237,35454),
       (1234,239999)]
deposit={}
for cust_id,amount in trans:
    if cust_id in deposit:
        deposit[cust_id].append(amount)
    else:
        deposit[cust_id]=[amount]
print(deposit)
```

```python
#costmer with maximum no of deposites
trans1 = {
    1237: [87522, 35454],
    1234: [1000, 239999],
    1999: [34657, 645578, 98765]
}

m=max(trans1.items(),key=lambda x:len(x[1]))  
print(m)                                                                                                            
```

```python
#who did maximum sum of deposites
trans1 = {
    1237: [87522, 35454],
    1234: [1000, 239999],
    1999: [34657, 645578, 98765]
}
m1=max(trans1.items(),key=lambda x:sum(x[1]))
print(m1)
```
