```python
#factorial
n=6
prod=1
for i in range(1,n+1):
    prod=prod*i
print(prod)
```

```python
def fact(n):
    product=1
    for i in range(1,n+1):
        product*=i
    return product
print(fact(6))
```

```python
#nCr=n!/(r!*(n-r)!)
def fact(x):
    prod=1
    for i in range(1,x+1):
        prod*=i
    return prod

def ncr(n,r):
    return fact(n)//(fact(r)*fact(n-r))
res=ncr(4,2)
print(res)
```
