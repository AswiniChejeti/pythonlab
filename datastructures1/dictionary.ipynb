```python
dict2={}
odd=list()
even=list()
for i in range(1,11):
    if i%2!=0:
        odd.append(i)
    else:
        even.append(i)
dict2.update({"odd":odd,"even":even})
print(dict2)
```

```python
email=['steven_A@gmail.com',
 'nick_A@outlook.com',
 'peter_C@gmail.com',
 'dan_B@outlook.com',
 'tim_A@gmail.com',
 'dave_C@rediff.com',
 'john_B@rediff.com',
 'jacob_A@gmail.com',
 'chris_D@yahoo.com']
dict3={}
for e in email:
    name,sec=e.split('@')[0].split('_')
    if sec not in dict3:
        dict3[sec]=[name]
    else:
        dict3[sec].append(name)
print(dict3)
```

```python
#print the service provider and how many users 
#print maximum number of users using service and min also
emails=['steven_A@gmail.com',
 'nick_A@outlook.com',
 'peter_C@gmail.com',
 'dan_B@outlook.com',
 'tim_A@gmail.com',
 'dave_C@rediff.com',
 'john_B@rew.com',
 'jacob_A@gmail.com',
 'chris_D@yahoo.com']
servicedict={}
for email in emails:
    service,_=email.split('@')[1].split('.')
    if service not in servicedict:
        servicedict[service]=1
    else:
        servicedict[service]+=1
print(servicedict)
#from the below comment we get only one service 
#_max=max(servicedict.items(), key=lambda x:x[1])[0]
#from the below code we get if maximum services are more than one
_max=max(servicedict.items(), key=lambda x:x[1])
s=[i for i in servicedict if servicedict[i]==_max[1]]

#_mnx=mix(servicedict.items(), key=lambda x:x[1])[0]
_min=min(servicedict.items(), key=lambda x:x[1])
w=[i for i in servicedict if servicedict[i]==_min[1]]
print(s)
print(w)

```

```python
#parent company nameis replaced by service provider
data=['steven_A@gmail.com',
 'nick_A@outlook.com',
 'peter_C@gmail.com',
 'dan_B@outlook.com',
 'tim_A@gmail.com',
 'dave_C@rediff.com',
 'john_B@rediff.com',
 'jacob_A@gmail.com',
 'chris_D@yahoo.com']
new={'gmail': 'Google', 'outlook': 'Microsoft'}
service_dict={}
for email1 in data:
    ser,_=email1.split('@')[1].split('.')
    new_ser=new.get(ser)
    if new_ser:
        ser=new_ser
    if ser not in service_dict:
        service_dict[ser]=1
    else:
        service_dict[ser]+=1
print(service_dict)
```
