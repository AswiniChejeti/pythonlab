```python
#5, Tasks 

sales=[
    ("laptop",45000),
    ("phone",20000),
    ("tablet",30000),
    ("laptop",47000),
    ("phone",22000)
]

#1, Create a dictionary with product name as key and list of prices
sales_list={}
for item,price in sales:
    if item not in sales_list:
        sales_list[item]=[]
    sales_list[item].append(price)
print(sales_list)

#2, Find the average price of each product
product={}
for item,price in sales_list.items():
    avg=sum(price)/len(price)
    product[item]=avg
print(product)

#3,Identify the product with highest average price
_max=max(product.items(), key=lambda x:x[1])
print(_max)
```

```python
#6, minimum 8 characters (atleast 1 uppercase letter,atleast 1 digit,atleast 1 special character (@#$%))

password="GenAI@2025"
uppercase=False
digit=False
specialcase=False
special="@#$%"
if len(password)<8:
    print("Invalid password")
else:
    for ch in password:
        if ch.isupper():
            uppercase=True
        elif ch.isdigit():
            digit=True
        elif ch in special:
            specialcase=True
    if uppercase and digit and specialcase:
        print("valid password") 
    else:
        print("Invalid password")   
```

```python
#8  1,Count uppercase letters

paragraph="GenAI 2025!\nPython is #1\nAI@Future"
count=0
for ch in paragraph:
    if ch.isupper():
        count+=1
print("upper case letters count in data  : ",count)

#2,Count special character 
scount=0
special=" !#@"
for ch in paragraph:
    if ch in special:
        scount+=1
print("special character count  : ",scount)

#3,Replace all digits with *
paragraph1=paragraph
result=""
for ch in paragraph1:
    if ch.isdigit():
        result+="*"
    else:
        result+=ch
print(result)

#4,Display the paragraph in reverse order(line-wise)
p=paragraph
l=p.split("\n")
for line in l:
    print(line[::-1])
```
