```python
#odd and even numbers in two sets
odd=set()
even=set()
for i in range(1,11):
    if i%2==0:
        even.add(i)
    else:
        odd.add(i)
print(odd)
print(even)
```

```python
primes=set()
odds=set()
output=set()
for i in range(1,11):
    if i%2!=0:
        odds.add(i)
    if i>1:
        for j in range(2,i):
            if i%j==0:
                break
        else:
            primes.add(i)
output=odds.intersection(primes)
print(odds)
print(primes)
print(output)
```
